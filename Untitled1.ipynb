```python
import numpy as np
from math import sqrt
from random import gauss
import matplotlib.pyplot as plt
def genera(x0,c,u,o,n):
  datos=np.zeros(n)
  datos[0]=x0
  delta=0.0001
  o_delta=sqrt(delta)
  for i in range(n-1):
      s=datos[i]
      t=0
      w=o_delta*gauss(0,1)
      s=s+c*(s-u)*delta+o*sqrt(s)*w
      t=t+delta
      datos[i+1]=s
  return datos
c=-.3
u=7
o=0.02
x0=u
n=30
datos=genera(x0,c,u,o,n)
np.savetxt('dat.txt', datos)
plt.figure(figsize=(4,3))
plt.plot(datos)
plt.show()

```
